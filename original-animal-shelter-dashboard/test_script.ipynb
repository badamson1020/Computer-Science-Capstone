```python
# Python Testing Script for the crud_module

# Import CRUD module
from crud_module import AnimalShelter

# Instantiate an instance of the class
# Username and password exist inside the "__init__" method for the crud_module 
shelter = AnimalShelter()

# Utilizes the create method to create a new record
new_animal = {
    "name": "Goddard",
    "type": "Robot Dog",
    "breed": "Paw-thon",
    "age": 5
}

insertion = shelter.create(new_animal)
print(f"Insertion successful: {insertion}")  # Will print True if insert was successful

# Utilizes the read method to return records
query = {"type": "Robot Dog"}  # Search for newly created record
robot_dogs = shelter.read(query)
print("Robot dogs found in the collection:")
for robot_dog in robot_dogs:
    print(robot_dog)

# A second test query to see if access is possible to pre-existing documents
query = {"animal_type": "Cat"}  # Search for all cats
cats = shelter.read(query)[:10]  # Limit results to the first 10 cats
print("The first 10 cats found in the collection:")
for cat in cats:
    print(cat)
    
# Utilizes the update method to update one pre-existing document
update_result = shelter.update({"name": "Goddard"}, {"$set": {"breed": "Paw-thon 1.0"}})
print(f"Number of documents updated: {update_result}")
print("After update:", shelter.read({"breed": "Paw-thon 1.0"}))

# Utilizing the update method to test if multiple documents can be updated at once
update_result = shelter.update({"name": "Goddard"}, {"$set": {"breed": "Paw-thon 2.0"}}, many=True)
print(f"Number of documents updated: {update_result}")
print("After update:", shelter.read({"name": "Goddard"}))

# Utilizes the delete method to delete one pre-existing document
delete_result = shelter.delete({"name": "Goddard"})
print(f"Number of documents deleted: {delete_result}")
print("After delete:", shelter.read({"name": "Goddard"}))

# Utilizing the delete method to test if multiple documents can be deleted at once
delete_result = shelter.delete({"name": "Goddard"}, many=True)
print(f"Number of documents deleted: {delete_result}")
print("After delete:", shelter.read({"name": "Goddard"}))
```
